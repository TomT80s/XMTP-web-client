```python
import subprocess
import sys
import csv
import time

# --- AUTO-INSTALLER ---
def install_requirements():
    required = {"web3", "requests"}
    for package in required:
        try:
            __import__(package)
        except ImportError:
            print(f"Installing {package}...")
            subprocess.check_call([sys.executable, "-m", "pip", "install", package])

install_requirements()

# Now we can safely import the libraries
from web3 import Web3
import requests

# --- CONFIGURATION ---
ALCHEMY_URL = "https://base-mainnet.g.alchemy.com/v2/WF22rihCN8lDVAy1XBBOXaw9iOtX33gH"
CONTRACT_ADDRESS = "0xe8D44EDC296CF804Bfb6D7ebb66d9aB9377D6Ad5"
IPFS_GATEWAY = "https://ipfs.io/ipfs/"
OUTPUT_FILE = "metadata.csv"

START_ID = 1
END_ID = 3389

ABI = '[{"inputs":[{"internalType":"uint256","name":"tokenId","type":"uint256"}],"name":"tokenURI","outputs":[{"internalType":"string","name":"","type":"string"}],"stateMutability":"view","type":"function"}]'

# --- SETUP ---
w3 = Web3(Web3.HTTPProvider(ALCHEMY_URL))
contract = w3.eth.contract(address=w3.to_checksum_address(CONTRACT_ADDRESS), abi=ABI)

def resolve_uri(uri):
    if uri.startswith("ipfs://"):
        return uri.replace("ipfs://", IPFS_GATEWAY)
    return uri

def process_mint_data():
    if not w3.is_connected():
        print("Error: Could not connect to Alchemy. Check your URL.")
        return

    print(f"Connected to Base via Alchemy. Processing {END_ID} tokens...")

    with open(OUTPUT_FILE, mode='w', newline='', encoding='utf-8') as file:
        writer = csv.writer(file)
        writer.writerow(["TokenID", "TokenURI", "Name", "Description", "Attributes", "Image_URL"])

        for token_id in range(START_ID, END_ID + 1):
            try:
                # 1. Get URI from Blockchain
                raw_uri = contract.functions.tokenURI(token_id).call()
                resolved_url = resolve_uri(raw_uri)

                # 2. Fetch JSON from IPFS
                name, desc, traits_str, image_url = "N/A", "N/A", "N/A", "N/A"
                try:
                    response = requests.get(resolved_url, timeout=10)
                    if response.status_code == 200:
                        data = response.json()
                        name = data.get('name', f"Token #{token_id}")
                        desc = data.get('description', "")
                        image_url = resolve_uri(data.get('image', ""))
                        traits = data.get('attributes', [])
                        traits_str = " | ".join([f"{t.get('trait_type', 'Trait')}: {t.get('value', 'N/A')}" for t in traits])
                except Exception:
                    pass # Keep moving if metadata fetch fails

                writer.writerow([token_id, raw_uri, name, desc, traits_str, image_url])

            except Exception as e:
                print(f"Skipping ID {token_id} due to error: {e}")

            if token_id % 50 == 0:
                print(f"Progress: {token_id}/{END_ID} tokens processed...")

            time.sleep(0.02) # Baseline delay for stability

    print(f"Success! Data saved to {OUTPUT_FILE}")

if __name__ == "__main__":
    process_mint_data()

files.download('metadata.csv')
```
